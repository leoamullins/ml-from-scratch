```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_regression
```

# Linear Regression

This linear regression method solves the least squares problem using the normal equation. Therefore this is a closed form solution and does not require gradient descent.

```python
X, y = make_regression(n_samples=100, n_features=1, noise=10, random_state=0)
```

```python
from mlscratch.linear_regression import LinearRegression

model = LinearRegression()

```

```python
model.fit(X, y)
```

```python
inputs = np.linspace(X.min(), X.max(), 100)
outputs = model.predict(inputs)

plt.scatter(X, y, label="data")
plt.plot(inputs, outputs, color="red", label="fit")   # the line
plt.legend()
plt.show()
```

# Polynomial Regression

Here we simply use linear regression on polynomial features. This is done with the `fit_poly` method, which creates the features to the `degree = 3` in this case. Polynomial regression is a linear regression problem.

```python
X = X.ravel()
y = X ** 3
```

```python
model_poly = LinearRegression()
model_poly.fit_poly(X, y, degree = 3)
```

```python
inputs = np.linspace(X.min(), X.max(), 100)
outputs = model_poly.predict_poly(inputs)

plt.scatter(X, y, label="data")
plt.plot(inputs, outputs, color="red", label="fit")   # the line
plt.legend()
```
