# Logistic Regression

```python
from mlscratch.logistic_regression import LogisticRegression, MultinomialRegression
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris, make_classification
from sklearn.metrics import accuracy_score
```

### Binary Classification
For the binary classification we use the `LogisticRegression` class.

To start we initialise `X` and `y` feature and target datasets respectively.

```python
X, y = make_classification(
    n_features=2, n_informative=2, n_redundant=0,
    n_clusters_per_class=1, random_state=0,
)
```

We then create an instance of the `LogisticRegression` here called `binary_model`, this has adjustable hyperparameters including `learning_rate`, `positive_threshold` and `num_epochs`. However, these are all set to defaults in this case.

```python
binary_model = LogisticRegression()
```

All the methods implemented in this library follow the standard `fit` and `predict` workflow. Since this is purely used as an example notebook for this method, the model is trained on the whole dataset and will predict on the same dataset. The evaluation will follow on a larger dataset.

```python
binary_model.fit(X, y)
```

```python
y_pred = binary_model.predict(X)
```

```python
plt.scatter(X[:, 0], X[:, 1], c= y)
```

### Multinomial Regression (Multiple Classes)

For multiple classes we use the `MultinomialRegression` class. The *iris* dataset from scikit-learn will be used here.

```python
X, y = load_iris(return_X_y=True)
```

```python
model = MultinomialRegression()
model.fit(X, y)
```

```python
y_pred = model.predict(X)
```

```python
accuracy_score(y_pred, y)
```

A similar workflow is followed for consistency.
