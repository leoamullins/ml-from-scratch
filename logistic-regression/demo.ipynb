# Logistic Regression

```python
import numpy as np
from model import LogisticRegression, sigmoid
import matplotlib.pyplot as plt
```

```python
def make_logistic_data(n=5000, d=10, signal=2.5, true_b=-0.5, seed=0):
    rng = np.random.default_rng(seed)
 
    X = rng.standard_normal((n, d))         
 
    true_w = rng.standard_normal(d)
    true_w *= signal / np.linalg.norm(true_w)   
 
    z = X @ true_w + true_b
    p = 1 / (1 + np.exp(-z))
    y = rng.binomial(1, p).astype(float)     
 
    return X, y, true_w, true_b

X, y, true_w, true_b = make_logistic_data()
true_w, true_b
```

```python
model = LogisticRegression(alpha = 0.05, num_steps = 10000, positive_threshold=0.7)
```

```python
model.fit(X, y)
```

```python
plt.plot(model.loss_history)
plt.xlabel("epoch")
plt.ylabel("loss")
plt.show()
```

```python
w = model_small.w
b = model_small.b

domain = np.linspace(-1.0, 1.0, 100)
func = sigmoid(w * domain + b)
```
