# Logistic Regression

```python
import numpy as np
from model import LogisticRegression
import matplotlib.pyplot as plt
```

```python
def make_logistic_data(n=5000, d=10, signal=2.5, true_b=-0.5, seed=0):
    rng = np.random.default_rng(seed)
 
    X = rng.standard_normal((n, d))         
 
    true_w = rng.standard_normal(d)
    true_w *= signal / np.linalg.norm(true_w)   
 
    z = X @ true_w + true_b
    p = 1 / (1 + np.exp(-z))
    y = rng.binomial(1, p).astype(float)     
 
    return X, y, true_w, true_b

X, y, true_w, true_b = make_logistic_data()
true_w, true_b
```

```python
model = LogisticRegression(alpha = 0.05, num_steps = 10000, positive_threshold=0.7)
```

```python
model.fit(X, y)
```

```python
plt.plot(model.loss_history)
plt.xlabel("epoch")
plt.ylabel("loss")
plt.show()
```

# Small 1D dataset for plotting

```python
def make_separable_data(n=10, margin=1.0, spread=2.0, seed=1):
    rng = np.random.default_rng(seed)
    n_neg, n_pos = n // 2, n - n // 2

    x_neg = -margin - rng.uniform(0, spread, n_neg)
    x_pos = margin + rng.uniform(0, spread, n_pos)

    X = np.concatenate([x_neg, x_pos]).reshape(-1, 1)
    y = np.concatenate([np.zeros(n_neg), np.ones(n_pos)])

    order = rng.permutation(n)
    return X[order], y[order]

X_small, y_small = make_separable_data()

plt.scatter(X_small[:, 0], y_small, c=y_small, cmap="bwr", edgecolor="k")
plt.axvline(0, linestyle="--", color="gray")
plt.xlabel("x")
plt.ylabel("y")
plt.yticks([0, 1])
plt.show()

X_small, y_small
```
